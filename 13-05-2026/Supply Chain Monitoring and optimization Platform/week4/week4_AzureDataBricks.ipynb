```python
from pyspark.sql.functions import *

# Read CSV
df = spark.read.csv(
    "/Volumes/hexa_db/default/supplychain/week4_orders.csv",
    header=True,
    inferSchema=True
)

```

```python
# Show original data
df.show()

```

```python
# Filter delayed orders
delayed_df = df.filter(df.is_delayed == 1)

# Show filtered data
delayed_df.show()

```

```python
# Save cleaned output as CSV
delayed_df.write.mode("overwrite") \
    .csv("/Volumes/hexa_db/default/supplychain/cleaned_orders_csv")

# Save cleaned output as Delta
delayed_df.write.format("delta") \
    .mode("overwrite") \
    .save("/Volumes/hexa_db/default/supplychain/cleaned_orders_delta")

```

```python
# Create SQL view
delayed_df.createOrReplaceTempView(
    "orders"
)

# Run basic SQL query
result = spark.sql("SELECT * FROM orders")

# Show query result
result.show()
```
