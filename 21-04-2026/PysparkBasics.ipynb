```python
from pyspark.sql import SparkSession
spark =SparkSession.builder.appName("PySpark Basics").getOrCreate()
print("Spark Session Created")
```

```python
print(spark.version)
```

```python
data = [
    ("vaibhav",15,"battter"),
    ("Hardik",30,"All rounder"),
    ("Bumrah",35,"Bowler")
]

columns=["Name","Age","Role"]

df=spark.createDataFrame(data,columns)
df.show()
```

```python
df.printSchema()
```

```python
df.select("Name").show()
```

```python
df.select("Name","Role").show()
```

```python
df.filter(df.Age>25).show()
```

```python
df.show()
```

```python
df.orderBy("Age").show()
```

```python
df.orderBy(df.Age.desc()).show()
```

```python
df.groupBy("Role").count().show()
```
