```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName("AbbasOviya").getOrCreate()
```

```python
data = [
    (1, "Rahul", "IT", 70000, "2024-01-10"),
    (2, "Sneha", "HR", 60000, "2024-02-15"),
    (3, "Arjun", "IT", 75000, "2024-03-01"),
    (4, "Priya", "Finance", 80000, "2024-01-25"),
    (5, "Karan", None, 50000, "2024-02-20")
]

columns = ["emp_id", "name", "department", "salary", "joining_date"]

df = spark.createDataFrame(data, columns)
df.show()
```

```python
df.groupBy("department").count().show()
```

```python
df.orderBy("salary").show()
```

```python
df.orderBy(df.salary.desc()).show()
```

```python
df.select("name","department","salary").show()
```

```python
df.select(df.salary>70000).show()
```

```python
dept_data = [
    ("IT", "Bangalore"),
    ("HR", "Mumbai"),
    ("Finance", "Delhi")
]

dept_columns = ["department", "location"]

dept_df = spark.createDataFrame(dept_data, dept_columns)
dept_df.show()
```

```python
df.show()
```

```python
df.join(dept_df,on="department",how="inner").show()
```

```python
df.join(dept_df,on="department",how="left").show()
```

```python
df.show()
```

```python
df.withColumn("Bonus",df.salary*0.10).show()
```

```python
df.filter(df.department.isNull()).show()
```

```python
df.na.fill({"department":"Unknown"}).show()
```

```python
df.na.drop().show()
```

```python
df.join(dept_df,on="department",how="left").show()
```

```python
df.show()
```

```python
from pyspark.sql.functions import lower,upper,length
df.select("name",upper("name").alias("upper_name"),lower("name").alias("lower_name"),length("name").alias("names_len")).show()
```
