```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName("myfirstTask").getOrCreate()
print("spark created")
```

```python
print("Add 5 employees")
data=[
    ("abbas",100000,"CSE"),
    ("oviya",85000,"IT"),
    ("murugan",50000,"AIDS"),
    ("siva",200000,"CSBS"),
    ("avinash",35000,"CSE")
]

column=["Name","Salary","Dept"]

df=spark.createDataFrame(data,column)
df.show()
```

```python
print("Show only Name and Salary columns")
df.select("Name","Salary").show()
```

```python
print("Filter employees with salary greater than 50000")
df.filter(df.Salary>50000).show()
```

```python
print("Count employees per department")
df.groupBy("Dept").count().show()
```

```python
print("Sort employees by salary")
df.orderBy("Salary").show()
```
