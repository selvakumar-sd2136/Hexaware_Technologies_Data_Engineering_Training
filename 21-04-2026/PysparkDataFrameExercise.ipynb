```python
df_dept=spark.read.csv("departments.csv",header=True,inferSchema=True)
df_dept.show()
```

```python
df_emply.select("name","salary").show()
```

```python
df_emply.filter(df_emply.joining_date>"2023-01-01").show()
```

```python
df_emply.filter(df_emply.salary>65000).show()
```

```python
df_emply.filter(df_emply.department=="IT").show()
```

```python
df_emply.filter((df_emply.department=="IT")&(df_emply.salary>75000)).show()
```

```python
df_emply.drop("joining_date")
```

```python
df_emply.orderBy("salary").show()
```

```python
df_emply.orderBy(df_emply.salary.desc()).show()
```

```python
df_emply.orderBy(df_emply.salary.desc()).limit(3).show()
```

```python
df_emply.selectExpr("sum(salary) as total_salary").show()
```

```python
df_emply.selectExpr("avg(salary) as Average_Salary").show()
```

```python
df_emply.selectExpr("max(salary) as Max_Salary").show()
```

```python
df_emply.selectExpr("min(salary) as Min_Salary").show()
```

```python
df_emply.groupBy("department").count().show()
```

```python
df_emply.groupBy("department").avg("salary").show()
```

```python
df_emply.groupBy("department").sum("salary").show()
```

```python
df_emply.join(df_dept,on="department",how="inner").show()
```

```python
df_emply.join(df_dept,on="department",how="inner").show()
```

```python
df_emply.join(df_dept,on="department",how="inner").select(df_emply.name,df_dept.location).show()
```

```python
df_emply.withColumn("Bonus",df_emply.salary*0.10).show()
```

```python
from pyspark.sql.functions import lit
df_emply.withColumn("Company",lit("BotCampus")).show()
```

```python
df_emply.filter(df_emply.department.isNull()).show()
```

```python
df_emply.fillna({"department": "Unknown"}).show()
```

```python
df_emply.dropna().show()
```

```python
from pyspark.sql.functions import upper,length

df_emply.withColumn("name_upper", upper("name")).show()
df_emply.withColumn("name_length", length("name")).show()
```

```python
from pyspark.sql.functions import col

df_emply.withColumn("salary_in_lakhs", col("salary") / 100000).show()
```
